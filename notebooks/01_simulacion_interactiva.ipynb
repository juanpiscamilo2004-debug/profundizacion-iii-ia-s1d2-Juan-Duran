# Simulación interactiva de una pelota de tenis de mesa

Este notebook permite explorar el modelo físico migrado desde MATLAB a Python.

Los parámetros se modifican mediante sliders y la simulación utiliza las
funciones implementadas en `src/table_tennis_sim/`.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output

from src.table_tennis_sim.parameters import TableTennisParameters
from src.table_tennis_sim.simulation import run_simulation
```

```python
def graficas(resultado):
    t = resultado.time

    fig, ax = plt.subplots(figsize=(10, 5))
    ax.plot(t, resultado.position[0] / 1000, label="X")
    ax.plot(t, resultado.position[1] / 1000, label="Y")
    ax.plot(t, resultado.position[2] / 1000, label="Z")
    ax.set_xlabel("Tiempo (s)")
    ax.set_ylabel("Posición (m)")
    ax.set_title("Posición vs tiempo")
    ax.legend()
    ax.grid()
    plt.show()

    fig, ax = plt.subplots(figsize=(10, 5))
    ax.plot(t, resultado.velocity[0] / 1000, label="Vx")
    ax.plot(t, resultado.velocity[1] / 1000, label="Vy")
    ax.plot(t, resultado.velocity[2] / 1000, label="Vz")
    ax.set_xlabel("Tiempo (s)")
    ax.set_ylabel("Velocidad (m/s)")
    ax.set_title("Velocidad vs tiempo")
    ax.legend()
    ax.grid()
    plt.show()

    fig, ax = plt.subplots(figsize=(10, 5))
    ax.plot(t, resultado.angular_velocity[0] / (2 * np.pi), label="ωx")
    ax.plot(t, resultado.angular_velocity[1] / (2 * np.pi), label="ωy")
    ax.plot(t, resultado.angular_velocity[2] / (2 * np.pi), label="ωz")
    ax.set_xlabel("Tiempo (s)")
    ax.set_ylabel("Velocidad angular (rev/s)")
    ax.set_title("Velocidad angular vs tiempo")
    ax.legend()
    ax.grid()
    plt.show()
```

```python
def explorar(
    vx=7000,
    vy=-3000,
    vz=-3000,
    omega_z=75,
    drag=2.7,
    magnus=0.01,
    table_restitution=0.77,
    net_restitution=0.5,
    table_friction=0.25,
    duration=1.5,
):
    parametros = TableTennisParameters(
        initial_velocity=(vx, vy, vz),
        initial_angular_velocity=(0, 0, omega_z * 2 * np.pi),
        drag=drag,
        magnus=magnus,
        table_restitution=table_restitution,
        net_restitution=net_restitution,
        table_friction=table_friction,
        duration=duration,
    )

    resultado = run_simulation(parametros)

    fig = plt.figure(figsize=(10, 7))
    ax = fig.add_subplot(111, projection="3d")

    ax.plot(
        resultado.position[0],
        resultado.position[1],
        resultado.position[2],
    )

    ax.set_xlabel("X (mm)")
    ax.set_ylabel("Y (mm)")
    ax.set_zlabel("Z (mm)")
    ax.set_title("Trayectoria de la pelota")

    plt.show()

    graficas(resultado)

    print(f"Muestras: {len(resultado.time)}")
    print(f"Duración: {resultado.time[-1]:.3f} s")
    print(f"Colisiones con mesa: {len(resultado.table_collision_indices)}")
    print(f"Colisiones con red: {len(resultado.net_collision_indices)}")

    print(
        "\nEfecto observado: los parámetros modificados cambian "
        "la trayectoria y/o la evolución de la velocidad y el giro."
    )
```

```python
widgets.interact(
    explorar,
    vx=widgets.IntSlider(
        value=7000,
        min=1000,
        max=10000,
        step=250,
        description="vx",
    ),
    vy=widgets.IntSlider(
        value=-3000,
        min=-6000,
        max=6000,
        step=250,
        description="vy",
    ),
    vz=widgets.IntSlider(
        value=-3000,
        min=-6000,
        max=2000,
        step=250,
        description="vz",
    ),
    omega_z=widgets.IntSlider(
        value=75,
        min=-150,
        max=150,
        step=5,
        description="omega z",
    ),
    drag=widgets.FloatSlider(
        value=2.7,
        min=0,
        max=5,
        step=0.1,
        description="Arrastre",
    ),
    magnus=widgets.FloatSlider(
        value=0.01,
        min=0,
        max=0.05,
        step=0.005,
        description="Magnus",
    ),
    table_restitution=widgets.FloatSlider(
        value=0.77,
        min=0,
        max=1,
        step=0.01,
        description="Rest. mesa",
    ),
    net_restitution=widgets.FloatSlider(
        value=0.5,
        min=0,
        max=1,
        step=0.01,
        description="Rest. red",
    ),
    table_friction=widgets.FloatSlider(
        value=0.25,
        min=0,
        max=1,
        step=0.01,
        description="Fricción",
    ),
    duration=widgets.FloatSlider(
        value=1.5,
        min=0.5,
        max=3,
        step=0.1,
        description="Duración",
    ),
);
```
